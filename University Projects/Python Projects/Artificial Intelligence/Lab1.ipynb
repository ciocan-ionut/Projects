```python
# Problema 1 - AI
max_word = ""
start = 0
n = len(text)

for i in range(n + 1):
    if i == n or text[i] == ' ':
        if i > start:
            current_word = text[start:i]
            if current_word > max_word:
                max_word = current_word
        start = i + 1
print(max_word)
# T: O(N) S: O(W) (unde W este lungimea celui mai lung cuvânt)
```

```python
# Problema 3 - EU
v1 = [1, 0, 2, 0, 3]
v2 = [1, 2, 0, 3, 1]
p = 0
min_len = min(len(v1), len(v2))
for i in range(0, min_len):
    if v1[i] != 0 and v2[i] != 0:
        p += v1[i] * v2[i]
print(p)
# T: O(N) S: O(N)
```

```python
# Problema 4 - EU
text2 = "ana are ana are mere rosii ana"
start = 0
n = len(text2)
text_dict = {}

for i in range(n + 1):
    if i == n or text2[i] == ' ':
        if i > start:
            current_word = text2[start:i]
            if current_word in text_dict:
                text_dict[current_word] += 1
            else:
                text_dict[current_word] = 1
        start = i + 1
for key, value in text_dict.items():
    if value == 1:
        print(key)
# T: O(N) S: O(U) (unde U e numarul de cuvinte unice din test)
```

```python
# Problema 4 - AI
from collections import Counter
from typing import Iterator

def extrage_cuvinte(text: str) -> Iterator[str]:
    """
    Generator care extrage cuvintele rând pe rând fără a crea
    o listă intermediară în memorie.
    """
    start = 0
    n = len(text)
    for i in range(n + 1):
        if i == n or text[i] == ' ':
            if i > start:
                yield text[start:i]
            start = i + 1

def cuvinte_aparitie_unica(text: str) -> list[str]:
    """
    Returnează o listă cu acele cuvinte care apar o singură dată.
    """
    # Counter consumă iteratorul direct, actualizând dicționarul de frecvențe.
    frecvente = Counter(extrage_cuvinte(text))

    # Extragem doar cuvintele a căror valoare (frecvență) este exact 1
    return [cuvant for cuvant, contor in frecvente.items() if contor == 1]

rezultat = cuvinte_aparitie_unica(text2)
print(rezultat)
# T: O(N) S: O(U) (unde U e numarul de cuvinte unice din test)
```

```python
# Problema 6 - AI
def element_majoritar(sir: list[int]) -> int:
    """
    Găsește elementul majoritar folosind algoritmul de votare Boyer-Moore.
    Presupune că șirul conține întotdeauna un element majoritar.
    """
    candidat = None
    contor = 0

    for numar in sir:
        # Dacă contorul este 0, alegem un nou candidat
        if contor == 0:
            candidat = numar
            contor = 1
        # Dacă numărul curent este același cu candidatul, creștem contorul
        elif numar == candidat:
            contor += 1
        # Dacă este diferit, scădem contorul
        else:
            contor -= 1

    return candidat
rezultat = element_majoritar(vmaj)
print(rezultat)
# T: O(N) S: O(1)
```

```python
# Problema 7 - AI
import random

def quickselect_k_maxim(sir: list[int], k: int) -> int:
    """
    Găsește al k-lea cel mai mare element dintr-un șir folosind Quickselect.
    Modifică șirul original in-place pentru a menține complexitatea spațială O(1).
    """
    n = len(sir)
    # Transformăm "al k-lea cel mai mare" în indexul corespunzător (0-indexed)
    index_cautat = n - k

    stanga = 0
    dreapta = n - 1

    while stanga <= dreapta:
        # Alegem un pivot aleatoriu pentru a evita cazul defavorabil O(N^2)
        pivot_idx = random.randint(stanga, dreapta)

        # Mutăm pivotul la final pentru a procesa restul elementelor
        sir[pivot_idx], sir[dreapta] = sir[dreapta], sir[pivot_idx]

        # Etapa de partiționare (Lomuto partition scheme)
        i = stanga
        for j in range(stanga, dreapta):
            if sir[j] < sir[dreapta]:
                sir[i], sir[j] = sir[j], sir[i]
                i += 1

        # Punem pivotul la locul lui definitiv
        sir[i], sir[dreapta] = sir[dreapta], sir[i]

        # Verificăm dacă pivotul a ajuns exact pe poziția căutată
        if i == index_cautat:
            return sir[i]
        elif i < index_cautat:
            # Elementul e în dreapta partiției
            stanga = i + 1
        else:
            # Elementul e în stânga partiției
            dreapta = i - 1

    return -1 # În caz de date invalide

rezultat = quickselect_k_maxim(vk, k)
print(rezultat)
# T: O(N) S: O(1)
```

```python
# Problema 8 - EU
def int_to_binary(nr):
    if nr == 0 or nr == 1:
        return str(nr)
    else:
        return int_to_binary(nr//2) + str(nr % 2)

n = 4
for i in range(1, n + 1):
    print(int_to_binary(i))
# T: O(NlogN) S: O(logN)
```

```python
#Problema 8 - AI
from typing import Iterator

def genereaza_binar(n: int) -> Iterator[str]:
    """
    Generator care produce reprezentarea binară a numerelor de la 1 la n.
    Folosește funcția nativă bin() și elimină prefixul '0b'.
    """
    for i in range(1, n + 1):
        # bin(i) returnează un string de forma '0b10',
        # așa că folosim [2:] pentru a tăia primele două caractere.
        yield bin(i)[2:]

# Exemplul dat
rezultat = list(genereaza_binar(n))
print(rezultat)
# T: O(NlogN) S: O(logN)
```

```python
# Problema 9 - EU
mat = [[0, 2, 5, 4, 1],
[4, 8, 2, 3, 7],
[6, 3, 4, 6, 2],
[7, 3, 1, 8, 3],
[1, 5, 7, 9, 4]]
lista = [[(1, 1), (3, 3)], [(2, 2), (4, 4)]]

for pereche in lista:
    suma = 0
    i = pereche[0][0]
    n = pereche[1][0]
    m = pereche[1][1]
    while i <= n:
        j = pereche[0][1]
        while j <= m:
            suma += mat[i][j]
            j += 1
        i += 1
    print(suma)
# T: O(K * N * M) S: O(1)
```

```python
# Problema 9 - AI
def construieste_sume_partiale(matrice: list[list[int]]) -> list[list[int]]:
    """
    Construiește matricea de sume parțiale 2D.
    """
    if not matrice or not matrice[0]:
        return []

    n = len(matrice)
    m = len(matrice[0])

    # Creăm o matrice (n+1) x (m+1) plină cu 0
    pref = [[0] * (m + 1) for _ in range(n + 1)]

    for i in range(n):
        for j in range(m):
            # Suma curentă = valoarea din matrice
            # + suma de deasupra + suma din stânga - intersecția (care a fost adunată de 2 ori)
            pref[i+1][j+1] = matrice[i][j] + pref[i][j+1] + pref[i+1][j] - pref[i][j]

    return pref

def suma_submatrice(pref: list[list[int]], r1: int, c1: int, r2: int, c2: int) -> int:
    """
    Calculează suma elementelor dintr-o sub-matrice în timp O(1).
    (r1, c1) - colțul stânga-sus; (r2, c2) - colțul dreapta-jos.
    """
    # Folosim indicii deplasați cu +1 din cauza rândului/coloanei de zerouri
    return pref[r2+1][c2+1] - pref[r1][c2+1] - pref[r2+1][c1] + pref[r1][c1]

def rezolva_interogari(matrice: list[list[int]], interogari: list[tuple]) -> list[int]:
    """
    Procesează o listă de interogări pe o matrice dată.
    """
    pref = construieste_sume_partiale(matrice)
    rezultate = []

    for (r1, c1), (r2, c2) in interogari:
        rezultate.append(suma_submatrice(pref, r1, c1, r2, c2))

    return rezultate

rezultate = rezolva_interogari(mat, lista)
print(rezultate)
# T: O(N * M + K) S: O(N * M)
```

```python
# Problema 10 - EU
mat = [[0,0,0,1,1],
[0,1,1,1,1],
[0,0,1,1,1]]

max = 0
linie = 0
for i in range(len(mat)):
    ones = mat[i].count(1)
    if ones > max:
        max = ones
        linie = i
print(linie + 1)
# T: O(N * M) S: O(1)
```

```python
# Problema 10 - AI
def linie_maxim_unu(matrice: list[list[int]]) -> int:
    """
    Găsește indexul liniei cu cele mai multe elemente de 1
    într-o matrice binară sortată crescător pe linii.
    Returnează -1 dacă matricea este goală sau conține doar 0.
    """
    if not matrice or not matrice[0]:
        return -1

    n = len(matrice)
    m = len(matrice[0])

    index_linie_max = -1

    # Pornim din colțul dreapta-sus
    i = 0          # rândul 0
    j = m - 1      # ultima coloană

    while i < n and j >= 0:
        if matrice[i][j] == 1:
            # Am găsit un 1. Această linie devine noul maxim.
            index_linie_max = i
            # Ne deplasăm la stânga pentru a căuta primul 1 din această secvență
            j -= 1
        else:
            # Am dat de un 0. Această linie nu ne mai poate oferi un număr
            # mai mare de 1 decât am găsit deja. Trecem la linia următoare.
            i += 1

    return index_linie_max
rezultat = linie_maxim_unu(mat)
print(rezultat + 1)
```
